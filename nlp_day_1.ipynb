```python
# Text Preprocessing
import nltk
nltk.download('punkt_tab')
```

```python
# tokenization
# word tokenization
from nltk.tokenize import word_tokenize
text = "NLP is very interesting domain happiness gathering studying listening hopping. nlp is the best subject in AI"
tokens = word_tokenize(text)
print(tokens)
```

```python
# sentence tokenization
from nltk.tokenize import sent_tokenize
text1 = "NLP is very interesting domain. NLP is used in chatbot"
tokens1 = sent_tokenize(text1)
print(tokens1)
```

```python
nltk.download('stopwords')
```

```python
from nltk.corpus import stopwords
stop_words = stopwords.words('english')
print(stop_words)
```

```python
# remove the stopwords
stop_words1 = set(stopwords.words('english'))
filter = [w for w in tokens if w.lower() not in stop_words1]
print(filter)
```

```python
# stemming
from nltk.stem import PorterStemmer
stemmer = PorterStemmer()
stemmed_words = [stemmer.stem(word) for word in filter]
print(stemmed_words)
```

```python
# lemmatization
nltk.download('wordnet')
nltk.download('omw-1.4')

from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()
lemmatized_words = [lemmatizer.lemmatize(word, pos='v') for word in filter ]
print(lemmatized_words)
```

```python
# bag of words
converted_text = " ".join(lemmatized_words)
print(converted_text)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
corpus = [converted_text]
vectorize = CountVectorizer(ngram_range=(1,3))  # ngram
vectorized_text = vectorize.fit_transform(corpus)
print(vectorize.get_feature_names_out())
print(vectorized_text.toarray())
```

```python
# tf-idf
from sklearn.feature_extraction.text import TfidfVectorizer
corpus1 = [converted_text]
vectorizer1 = TfidfVectorizer()
vectorized_text1 = vectorizer1.fit_transform(corpus1)
print(vectorizer1.get_feature_names_out())
print(vectorized_text1.toarray())
```

```python
# word2vec

sentences = [
    ["nlp", "is", "Interesting"],
    ["machine", "learning","is","powerful"],
    ["deep","learning","is","advanced"],
    ["nlp","and","machine","learning"],
    ["i","love","studying","nlp"],
    ["students","prefer","machine","learning"],
    ["nlp","helps","in","text","processing"],
    ["learning","new","'topics","is","fun"],
    ["artificial","Intelligence","is","future"]
]
```

```python
from gensim.models import Word2Vec
model = Word2Vec(
    sentences,
    vector_size=50,
    window=3,
    min_count=1,
    workers=700
)
```

```python
print("Vector for nlp")
print(model.wv['nlp'])
```

```python
print("\nwords similar to nlp")
print(model.wv.most_similar('nlp'))
```

```python
# vocabulary
print("Vocabulary: ")
print(model.wv.index_to_key)
```

```python
# sentence to vector
import numpy as np
def sent_vec(sentence):
  words = sentence.lower().split()
  vectors = []
  for word in words:
    if word in model.wv:
      vectors.append(model.wv[word])
  return np.mean(vectors,axis=0)

sentence = "NLP is interesting"
vec = sent_vec(sentence)
print("sentence vector: ")
print(vec)
```

```python
# Cosine similarity score
from sklearn.metrics.pairwise import cosine_similarity

s1 = sent_vec("nlp is interesting")
s2 = sent_vec("ai learnig is interesting")
similarity = cosine_similarity([s1],[s2])
print("sentence similarity\n")
print(similarity[0][0])
```

```python
import numpy as np
# Text classification
from sklearn.linear_model import LogisticRegression

x = [
    sent_vec("nlp is interesting"),
    sent_vec("machine learning is interesting"),
    sent_vec("I hate studying nlp"),
    sent_vec("deep learning is boring")
]
y = [1,1,0,0]

model_1 = LogisticRegression()
model_1.fit(x,y)

test = sent_vec("i don't like nlp")
print("Predictions: ", model_1.predict([test]))
```

```python
from sklearn.cluster import KMeans
x = [
    sent_vec("nlp is interesting"),
    sent_vec("machine learning is interesting"),
    sent_vec("I hate studying nlp"),
    sent_vec("deep learning is boring")
]

kmeans = KMeans(n_clusters=2)
kmeans.fit(x)
print("Cluster: ", kmeans.labels_)
```

## N-grams

N-grams are contiguous sequences of N items from a given sample of text or speech. The items can be characters, syllables, or words. N-grams are widely used in natural language processing (NLP) for various tasks such as language modeling, text classification, spell checking, and machine translation.

Here are some common types of n-grams:

*   **Unigrams (1-gram):** Individual words (e.g., "the", "cat", "sat").
*   **Bigrams (2-gram):** Pairs of adjacent words (e.g., "the cat", "cat sat", "sat on").
*   **Trigrams (3-gram):** Triplets of adjacent words (e.g., "the cat sat", "cat sat on").

### How N-grams are used:

1.  **Language Modeling:** Predicting the next word in a sequence based on the preceding `N-1` words.
2.  **Text Classification:** Features for classifying text documents.
3.  **Spell Correction:** Identifying probable misspellings by checking if sequences of characters or words are common.
4.  **Information Retrieval:** Matching queries to documents.
5.  **Machine Translation:** Evaluating the fluency of translated text.

```python
from nltk.util import ngrams
from nltk.tokenize import word_tokenize

# Example text
text_for_ngram = "NLP is a fascinating field. It helps in understanding and processing human language."

# Tokenize the text into words
words_for_ngram = word_tokenize(text_for_ngram.lower())

print(f"Original text: {text_for_ngram}")
print(f"Tokenized words: {words_for_ngram}\n")

# Generate unigrams (1-grams)
unigrams = list(ngrams(words_for_ngram, 1))
print(f"Unigrams (1-grams): {unigrams}\n")

# Generate bigrams (2-grams)
bigrams = list(ngrams(words_for_ngram, 2))
print(f"Bigrams (2-grams): {bigrams}\n")

# Generate trigrams (3-grams)
trigrams = list(ngrams(words_for_ngram, 3))
print(f"Trigrams (3-grams): {trigrams}")
```

## N-grams for Text Classification

N-grams can be used as features to train a text classification model. Here, we'll demonstrate a simple example using `CountVectorizer` to extract bigram features from a small corpus and then train a `LogisticRegression` model.

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.linear_model import LogisticRegression
from nltk.tokenize import word_tokenize

# 1. Sample Dataset
# Let's classify sentences as positive (1) or negative (0) sentiment
corpus_classification = [
    "This movie is great and fantastic",
    "I love this amazing film",
    "What a terrible movie, it's boring",
    "I hate this film, it's awful"
]
labels_classification = [1, 1, 0, 0]

# 2. Preprocessing (Tokenization and N-gram generation)
# We'll use bigrams (n-gram of 2 words) for this example

# You might need to preprocess further (like removing stopwords, stemming/lemmatization)
# For simplicity, we'll just tokenize and generate n-grams.

# Tokenize and rejoin for CountVectorizer
processed_corpus = []
for doc in corpus_classification:
    tokens_clf = word_tokenize(doc.lower())
    processed_corpus.append(" ".join(tokens_clf))

# Create CountVectorizer with bigram range
vectorizer_ngram = CountVectorizer(ngram_range=(1, 2)) # Using unigrams and bigrams
X_ngram = vectorizer_ngram.fit_transform(processed_corpus)

print("Features (N-grams):\n", vectorizer_ngram.get_feature_names_out())
print("\nVectorized N-gram features for corpus:\n", X_ngram.toarray())
```

```python
# 3. Train a Logistic Regression Classifier
model_ngram = LogisticRegression()
model_ngram.fit(X_ngram, labels_classification)

# 4. Predict on a new sentence
new_sentence = "This film is amazing"

# Preprocess the new sentence the same way
processed_new_sentence = " ".join(word_tokenize(new_sentence.lower()))

# Transform the new sentence using the *same* vectorizer
X_new_ngram = vectorizer_ngram.transform([processed_new_sentence])

prediction_ngram = model_ngram.predict(X_new_ngram)

print(f"\nNew sentence: '{new_sentence}'")
print(f"Predicted sentiment (1=Positive, 0=Negative): {prediction_ngram[0]}")
```

```python
# advanced natural language processing
processed_docs = [
    ['foods','delicious','restuarant','great','service'],
    ['love','eating','pizza','pasta'],
    ['stock','market','creashed','economic','slowdown'],
    ['investors','worried','market','volatility'],
    ['chef','prepared','wonderful','meal'],
    ['stocks','bonds','important','investment']
]

from gensim import corpora
from gensim.models import LdaModel

dictionary = corpora.Dictionary(processed_docs) # each word get an unique numerical id
corpus = [dictionary.doc2bow(doc) for doc in processed_docs]  # it gives word_id , frequency
lda_model = LdaModel(
    corpus= corpus,
    id2word= dictionary, # mapping word id
    num_topics= 2,       # no of words
    random_state= 42,
    passes= 10           # no of iterations
)

topics = lda_model.print_topics(num_words=6)
for topic in topics:
  print(topic)

for i,doc in enumerate(corpus):
  print(f"Document {i}: ", lda_model[doc])
```

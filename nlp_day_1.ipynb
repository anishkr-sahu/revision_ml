```python
# Text Preprocessing
import nltk
nltk.download('punkt_tab')
```

```python
# tokenization
# word tokenizatio
from nltk.tokenize import word_tokenize
text = "NLP is very interesting domain happiness gathering studying listening hopping"
tokens = word_tokenize(text)
print(tokens)
```

```python
# sentence tokenization
from nltk.tokenize import sent_tokenize
text1 = "NLP is very interesting domain. NLP is used in chatbot"
tokens1 = sent_tokenize(text1)
print(tokens1)
```

```python
nltk.download('stopwords')
```

```python
from nltk.corpus import stopwords
stop_words = stopwords.words('english')
print(stop_words)
```

```python
# remove the stopwords
stop_words1 = set(stopwords.words('english'))
filter = [w for w in tokens if w.lower() not in stop_words1]
print(filter)
```

```python
# stemming
from nltk.stem import PorterStemmer
stemmer = PorterStemmer()
stemmed_words = [stemmer.stem(word) for word in filter]
print(stemmed_words)
```

```python
# lemmatization
nltk.download('wordnet')
nltk.download('omw-1.4')

from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()
lemmatized_words = [lemmatizer.lemmatize(word, pos='v') for word in filter ]
print(lemmatized_words)
```
